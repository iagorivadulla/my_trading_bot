```python
from main import trading_monkey
from src.account import Account
from src.markets import Market
```

```python
market = Market('../data/market_data.parquet')
gains = []

for i in range(100):
    monkey = Account(5000)
    trading_monkey(monkey, market)

    last_prices = market.market.iloc[-1]
    equity = monkey.fiat + sum(a['amount'] * last_prices.Close[t] for t, a in monkey.assets.items())

    gains.append(equity)
    market.reset()
```

```python
import numpy as np
print(f'Median: {np.median(gains)}')
print(f'STD: {np.std(gains)}')
print(f'Mean: {np.mean(gains)}')
print(f'Min: {np.min(gains)}')
print(f'Max: {np.max(gains)}')
print(f'{len(gains)}')
```

```python
close = market.market['Close']
bh = (close.iloc[-1] / close.iloc[0]).mean() * 5000
print(bh, bh / 5000)
```
