```python
from src.markets import Market, read_market
from src.account import Account
from random_monkeys import trading_monkey
from src.bucks_every_month import fifty_bucks_every_month
```

```python
market = Market('../data/market_data_KO.parquet')
gains = []

for i in range(100):
    monkey = Account(50)
    trading_monkey(monkey, market)

    last_prices = market.market.iloc[-1]
    equity = monkey.fiat + sum(a['amount'] * last_prices.Close[t] for t, a in monkey.assets.items())

    gains.append(equity)
    market.reset()
```

```python
import numpy as np
print(f'Median: {np.median(gains)}')
print(f'STD: {np.std(gains)}')
print(f'Mean: {np.mean(gains)}')
print(f'Min: {np.min(gains)}')
print(f'Max: {np.max(gains)}')
print(f'{len(gains)}')
```
